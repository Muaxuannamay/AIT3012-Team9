```python
import mne
from mne_bids import BIDSPath, read_raw_bids
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline

bids_root = "./data" 

bids_path = BIDSPath(
    subject="01",
    task="oddball",   
    datatype="eeg",
    root=bids_root
)

raw = read_raw_bids(bids_path=bids_path, verbose=False)
raw.load_data()

print(raw.info)
```

```python
# Bản sao dữ liệu gốc lọc nhẹ high-pass 0.5 Hz để khử trôi nền DC
raw_base = raw.copy().filter(l_freq=0.5, h_freq=None, fir_design='firwin', verbose=False)

# LỌC KHÍA ĐIỆN LƯỚI 50 Hz (Chắn dải 48 - 52 Hz theo chuẩn Châu Âu)
raw_notch_50 = raw_base.copy()
raw_notch_50.notch_filter(
    freqs=[50.0, 100.0, 150.0], # Cắt 50 Hz và các sóng hài bậc cao
    notch_widths=4.0,           # Độ rộng dải chặn 4 Hz (tương đương chắn dải 48 - 52 Hz)
    method='fir',
    fir_design='firwin',
    verbose=False
)

# CHẮN DẢI 1 - 4 Hz (Triệt tiêu nhiễu chuyển động mắt chậm Slow EOG - để đối chiếu)
raw_bandstop_eog = raw_base.copy()
raw_bandstop_eog.filter(
    l_freq=4.0, 
    h_freq=1.0, 
    method='fir',
    fir_design='firwin',
    verbose=False
)
```

```python
chọn_kênh = ['Fp1']

fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

# Phổ tần số trước khi lọc Notch
raw_base.compute_psd(picks=chọn_kênh, fmin=1.0, fmax=120.0, verbose=False).plot(
    axes=axes[0], color='gray', show=False
)
axes[0].set_title(f"1. Phổ tần số (PSD) TRƯỚC khi lọc Notch (Kênh {chọn_kênh[0]})")
axes[0].axvline(50, color='red', linestyle='--', alpha=0.6, label='Nhiễu điện 50 Hz')
axes[0].legend()

# Phổ tần số sau khi lọc Notch 48-52 Hz
raw_notch_50.compute_psd(picks=chọn_kênh, fmin=1.0, fmax=120.0, verbose=False).plot(
    axes=axes[1], color='green', show=False
)
axes[1].set_title(f"2. Phổ tần số (PSD) SAU khi lọc Notch (48 - 52 Hz)")
axes[1].axvline(50, color='red', linestyle='--', alpha=0.6, label='Vết khía (Notch) 50 Hz')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
kênh = 'Fp1'
start_sample = 0
stop_sample = 2000

d_base = raw_base.get_data(picks=[kênh], start=start_sample, stop=stop_sample)[0] * 1e6
d_notch50 = raw_notch_50.get_data(picks=[kênh], start=start_sample, stop=stop_sample)[0] * 1e6
d_stop_eog = raw_bandstop_eog.get_data(picks=[kênh], start=start_sample, stop=stop_sample)[0] * 1e6
times = raw.times[start_sample:stop_sample]

plt.figure(figsize=(18, 7))
plt.plot(times, d_base, color='yellow', alpha=0.6, label='1. Tín hiệu nền (Chưa lọc Notch - Có gợn sóng răng cưa 50 Hz)')
plt.plot(times, d_notch50, color='green', linewidth=1.5, label='2. Lọc khía 48-52 Hz (Mượt hơn do mất nhiễu 50 Hz, bảo toàn nguyên vẹn sóng não)')
plt.plot(times, d_stop_eog, color='crimson', linewidth=1.2, linestyle='--', label='3. Chắn dải 1-4 Hz (Ép xẹp đỉnh chớp mắt nhưng làm biến dạng/mất sóng Delta)')

plt.title(f"So sánh tác động của Lọc chắn dải 48-52 Hz (Điện lưới) và 1-4 Hz (Slow EOG) trên kênh {kênh}")
plt.xlabel("Thời gian (giây)")
plt.ylabel("Biên độ (µV)")
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 1. Lấy dữ liệu toàn cục của toàn bộ 127 kênh qua 80 giây
data_orig_all = raw_base.get_data()
data_notch_all = raw_notch_50.get_data()
data_stop_eog_all = raw_bandstop_eog.get_data()

# % Năng lượng bị mất
power_orig = np.var(data_orig_all)
power_notch = np.var(data_notch_all)
power_eog = np.var(data_stop_eog_all)

loss_power_notch = (power_orig - power_notch) / power_orig * 100
loss_power_eog = (power_orig - power_eog) / power_orig * 100

# % Biên độ hiệu dụng
rms_orig = np.sqrt(np.mean(data_orig_all**2))
rms_notch = np.sqrt(np.mean(data_notch_all**2))
rms_eog = np.sqrt(np.mean(data_stop_eog_all**2))

loss_rms_notch = (rms_orig - rms_notch) / rms_orig * 100
loss_rms_eog = (rms_orig - rms_eog) / rms_orig * 100

print(f"   - Năng lượng (48-52 Hz) bị mất : {loss_power_notch:.4f} %")
print(f"   - Biên độ RMS (48-52 Hz) bị mất: {loss_rms_notch:.4f} %\n")

print(f"   - Năng lượng (1-4 Hz) bị mất  : {loss_power_eog:.4f} %")
print(f"   - Biên độ RMS (1-4 Hz) bị mất  : {loss_rms_eog:.4f} %")
```
