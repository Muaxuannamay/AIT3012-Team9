```python
import mne
from mne_bids import BIDSPath, read_raw_bids
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline

bids_root = "./data" 

bids_path = BIDSPath(
    subject="01",
    task="oddball",   
    datatype="eeg",
    root=bids_root
)

raw = read_raw_bids(bids_path=bids_path, verbose=False)
raw.load_data()

print(raw.info)
```

```python
raw_orig = raw.copy()

# Dải rộng (1 - 50 Hz): Tiền xử lý EEG tổng quát
raw_bp_broad = raw.copy()
raw_bp_broad.filter(l_freq=1.0, h_freq=50.0, fir_design='firwin', verbose=False)

# Dải hẹp (8 - 30 Hz): Cô lập nhịp Mu và Beta
raw_bp_narrow = raw.copy()
raw_bp_narrow.filter(l_freq=8.0, h_freq=30.0, fir_design='firwin', verbose=False)
```

```python
chọn_kênh = 'Fp1' 

start_sample = 0
stop_sample = 3000 

data_orig = raw_orig.get_data(picks=[chọn_kênh], start=start_sample, stop=stop_sample)[0]
data_broad = raw_bp_broad.get_data(picks=[chọn_kênh], start=start_sample, stop=stop_sample)[0]
data_narrow = raw_bp_narrow.get_data(picks=[chọn_kênh], start=start_sample, stop=stop_sample)[0]

times = raw.times[start_sample:stop_sample]

plt.figure(figsize=(20, 10))

plt.subplot(3, 1, 1)
plt.plot(times, data_orig, color='gray')
plt.title(f"1. Tín hiệu GỐC (Kênh {chọn_kênh})")
plt.ylabel("Biên độ (V)")
plt.grid(True, alpha=0.3)

plt.subplot(3, 1, 2)
plt.plot(times, data_broad, color='blue')
plt.title(f"2. Band-pass Dải rộng (1 - 50 Hz) - Loại bỏ trôi nền (<1 Hz) và nhiễu cao tần (>50 Hz)")
plt.ylabel("Biên độ (V)")
plt.grid(True, alpha=0.3)

plt.subplot(3, 1, 3)
plt.plot(times, data_narrow, color='darkorange')
plt.title(f"3. Band-pass Dải hẹp (8 - 30 Hz) - Cô lập nhịp Mu và Beta")
plt.xlabel("Thời gian (giây)")
plt.ylabel("Biên độ (V)")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
data_orig_all = raw_orig.get_data()
data_broad_all = raw_bp_broad.get_data()
data_narrow_all = raw_bp_narrow.get_data()

# % Năng lượng bị mất
power_orig = np.var(data_orig_all)
power_broad = np.var(data_broad_all)
power_narrow = np.var(data_narrow_all)

loss_power_broad = (power_orig - power_broad) / power_orig * 100
loss_power_narrow = (power_orig - power_narrow) / power_orig * 100

# % Biên độ hiệu dụng bị mất
rms_orig = np.sqrt(np.mean(data_orig_all**2))
rms_broad = np.sqrt(np.mean(data_broad_all**2))
rms_narrow = np.sqrt(np.mean(data_narrow_all**2))

loss_rms_broad = (rms_orig - rms_broad) / rms_orig * 100
loss_rms_narrow = (rms_orig - rms_narrow) / rms_orig * 100

print(f"   - Năng lượng (1-50 Hz) bị mất : {loss_power_broad:.6f} %")
print(f"   - Biên độ RMS (1-50 Hz) bị mất: {loss_rms_broad:.6f} %\n")

print(f"   - Năng lượng (8-30 Hz) bị mất : {loss_power_narrow:.6f} %")
print(f"   - Biên độ RMS (8-30 Hz) bị mất: {loss_rms_narrow:.6f} %")
```

```python
stop_sample = 10000
# Đổi đơn vị từ Volt (V) sang Microvolt (µV) 
orig_uV = data_orig * 1e6
broad_uV = data_broad * 1e6
narrow_uV = data_narrow * 1e6

plt.figure(figsize=(18, 10))

orig_uV_shifted = orig_uV - orig_uV[0]

plt.subplot(2, 1, 2)
plt.plot(times, orig_uV_shifted, color='gray', alpha=0.7, label='1. Gốc (Đã dịch điểm đầu về 0 µV)', linewidth=1.5)
plt.plot(times, broad_uV, color='blue', alpha=0.8, label='2. Band-pass Dải rộng (1 - 50 Hz)', linewidth=1.5)
plt.plot(times, narrow_uV, color='darkorange', alpha=0.9, label='3. Band-pass Dải hẹp (8 - 30 Hz) - Biên độ cực nhỏ', linewidth=1.5)

plt.title(f"Góc nhìn 2: So sánh biên độ dao động trên cùng tỷ lệ trục Y (Kênh {chọn_kênh})")
plt.xlabel("Thời gian (giây)")
plt.ylabel("Biên độ dao động (µV)")
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
