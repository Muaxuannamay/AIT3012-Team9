```python
import mne
from mne_bids import BIDSPath, read_raw_bids
import matplotlib.pyplot as plt
import numpy as np
from mne.viz import plot_compare_evokeds

%matplotlib inline

bids_root = "./data" 

bids_path = BIDSPath(
    subject="01",
    task="oddball",   
    datatype="eeg",
    root=bids_root
)

raw = read_raw_bids(bids_path=bids_path, verbose=False)
raw.load_data()

print(raw.info)
```

```python
raw_orig = raw.copy()

raw_01Hz = raw.copy()
raw_01Hz.filter(l_freq=0.1, h_freq=None, fir_design='firwin', verbose=False)

raw_05Hz = raw.copy()
raw_05Hz.filter(l_freq=0.5, h_freq=None, fir_design='firwin', verbose=False)
```

```python
chọn_kênh = 'Fp1' 

# Lấy dữ liệu 30 giây đầu tiên (30 giây * 1000Hz = 30,000 điểm)
start_sample = 0
stop_sample = 30000 

data_orig = raw_orig.get_data(picks=[chọn_kênh], start=start_sample, stop=stop_sample)[0]
data_01 = raw_01Hz.get_data(picks=[chọn_kênh], start=start_sample, stop=stop_sample)[0]
data_05 = raw_05Hz.get_data(picks=[chọn_kênh], start=start_sample, stop=stop_sample)[0]
times = raw.times[start_sample:stop_sample]


plt.figure(figsize=(20, 15))


# Đồ thị 1: Bản gốc
plt.subplot(3, 1, 1)
plt.plot(times, data_orig, color='gray')
plt.title(f"1. Tín hiệu GỐC (Kênh {chọn_kênh})")
plt.ylabel("Biên độ (µV)")
plt.grid(True, alpha=0.3)

# Đồ thị 2: Lọc 0.1 Hz
plt.subplot(3, 1, 2)
plt.plot(times, data_01, color='blue')
plt.title(f"2. Lọc High-pass 0.1 Hz (Cho ERP)")
plt.ylabel("Biên độ (µV)")
plt.grid(True, alpha=0.3)

# Đồ thị 3: Lọc 0.5 Hz
plt.subplot(3, 1, 3)
plt.plot(times, data_05, color='red')
plt.title(f"3. Lọc High-pass 0.5 Hz (Cho EEG liên tục)")
plt.xlabel("Thời gian (giây)")
plt.ylabel("Biên độ (µV)")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(70, 50))

plt.subplot(3, 1, 1)
plt.plot(times, data_orig, color='yellow', alpha=0.6, label='Gốc')
plt.plot(times, data_01, color='blue', linewidth=1, label='Lọc 0.1Hz')
plt.plot(times, data_05, color='red', linewidth=1, label='Lọc 0.5Hz')
plt.legend()
plt.grid(True, alpha=0.3)
```

```python
data_orig_all = raw_orig.get_data()
data_01_all = raw_01Hz.get_data()
data_05_all = raw_05Hz.get_data()

power_orig = np.var(data_orig_all)
power_01 = np.var(data_01_all)
power_05 = np.var(data_05_all)

loss_power_01 = (power_orig - power_01) / power_orig * 100
loss_power_05 = (power_orig - power_05) / power_orig * 100

rms_orig = np.sqrt(np.mean(data_orig_all**2))
rms_01 = np.sqrt(np.mean(data_01_all**2))
rms_05 = np.sqrt(np.mean(data_05_all**2))

loss_rms_01 = (rms_orig - rms_01) / rms_orig * 100
loss_rms_05 = (rms_orig - rms_05) / rms_orig * 100

print(f"   - Năng lượng (ERP P300) bị mất: {loss_power_01:.2f} %")
print(f"   - Biên độ (ERP P300) bị mất: {loss_rms_01:.2f} %")

print(f"   - Năng lượng (EEG) bị mất: {loss_power_05:.2f} %")
print(f"   - Biên độ (EEG) bị mất: {loss_rms_05:.2f} %")
```
