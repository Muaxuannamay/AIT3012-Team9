```python
import mne
from mne_bids import BIDSPath, read_raw_bids
import matplotlib.pyplot as plt
import numpy as np
from mne.viz import plot_compare_evokeds

%matplotlib inline

bids_root = "./data" 

bids_path = BIDSPath(
    subject="01",
    task="oddball",   
    datatype="eeg",
    root=bids_root
)

raw = read_raw_bids(bids_path=bids_path, verbose=False)
raw.load_data()

print(raw.info)
```

```python
import matplotlib.pyplot as plt
import numpy as np

raw_orig = raw.copy()

# Bản Low-pass 30 Hz
raw_30Hz = raw.copy()
raw_30Hz.filter(l_freq=None, h_freq=30.0, fir_design='firwin', verbose=False)

# Bản Low-pass 100 Hz
raw_100Hz = raw.copy()
raw_100Hz.filter(l_freq=None, h_freq=100.0, fir_design='firwin', verbose=False)
```

```python
chọn_kênh = 'Fp1' 

# Lấy dữ liệu 2 giây đầu tiên
start_sample = 0
stop_sample = 2000 

data_orig = raw_orig.get_data(picks=[chọn_kênh], start=start_sample, stop=stop_sample)[0]
data_30 = raw_30Hz.get_data(picks=[chọn_kênh], start=start_sample, stop=stop_sample)[0]
data_100 = raw_100Hz.get_data(picks=[chọn_kênh], start=start_sample, stop=stop_sample)[0]

times = raw.times[start_sample:stop_sample]

plt.figure(figsize=(20, 10))

# Đồ thị 1: Bản gốc
plt.subplot(3, 1, 1)
plt.plot(times, data_orig, color='gray')
plt.title(f"1. Tín hiệu GỐC (Kênh {chọn_kênh})")
plt.ylabel("Biên độ (µV)")
plt.grid(True, alpha=0.3)

# Đồ thị 2: Low-pass 100 Hz
plt.subplot(3, 1, 2)
plt.plot(times, data_100, color='green')
plt.title(f"2. Low-pass 100 Hz")
plt.ylabel("Biên độ (µV)")
plt.grid(True, alpha=0.3)

# Đồ thị 3: Low-pass 30 Hz
plt.subplot(3, 1, 3)
plt.plot(times, data_30, color='purple')
plt.title(f"3. Low-pass 30 Hz")
plt.xlabel("Thời gian (giây)")
plt.ylabel("Biên độ (µV)")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
data_orig_all = raw_orig.get_data()
data_30_all = raw_30Hz.get_data()
data_100_all = raw_100Hz.get_data()

# % Năng lượng bị mất
power_orig = np.var(data_orig_all)
power_30 = np.var(data_30_all)
power_100 = np.var(data_100_all)

loss_power_30 = (power_orig - power_30) / power_orig * 100
loss_power_100 = (power_orig - power_100) / power_orig * 100

# % Biên độ bị mất
rms_orig = np.sqrt(np.mean(data_orig_all**2))
rms_30 = np.sqrt(np.mean(data_30_all**2))
rms_100 = np.sqrt(np.mean(data_100_all**2))

loss_rms_30 = (rms_orig - rms_30) / rms_orig * 100
loss_rms_100 = (rms_orig - rms_100) / rms_orig * 100

print(f"   - Năng lượng (30 Hz) bị mất: {loss_power_30:.6f} %")
print(f"   - Biên độ (30 Hz) bị mất: {loss_rms_30:.6f} %")

print(f"   - Năng lượng (100 Hz) bị mất: {loss_power_100:.6f} %")
print(f"   - Biên độ (100 Hz) bị mất: {loss_rms_100:.6f} %")
```

```python
psds_orig, freqs = raw_orig.compute_psd(picks=['Fp1'], fmax=120).get_data(return_freqs=True)
psds_30, _ = raw_30Hz.compute_psd(picks=['Fp1'], fmax=120).get_data(return_freqs=True)
psds_100, _ = raw_100Hz.compute_psd(picks=['Fp1'], fmax=120).get_data(return_freqs=True)

plt.figure(figsize=(10, 6))
plt.semilogy(freqs, psds_orig[0], label='Gốc', alpha=0.7)
plt.semilogy(freqs, psds_100[0], label='Low-pass 100 Hz', alpha=0.7)
plt.semilogy(freqs, psds_30[0], label='Low-pass 30 Hz', alpha=0.7)
plt.axvline(30, color='purple', linestyle='--', alpha=0.5, label='Ngưỡng 30 Hz')
plt.axvline(100, color='green', linestyle='--', alpha=0.5, label='Ngưỡng 100 Hz')
plt.xlabel('Tần số (Hz)')
plt.ylabel('Mật độ phổ công suất (µV²/Hz)')
plt.title('Phổ công suất tín hiệu EEG (Kênh Fp1)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
"""
Biểu đồ này được vẽ với tác dụng chứng minh rằng có sự mất mát năng lượng xảy ra, dù rất nhỏ :))
Do phần lớn năng lượng của tín hiệu EEG tập trung ở các tần số thấp, và năng lượng giảm dần khi tần số tăng lên
"""
```
